```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_excel("./milan_rent_clean.xlsx")
data.head()
```

```python
plt.hist(data["surface_n"],bins=40)
```

```python
total_listings = len(data["surface_n"])
print("Number of Listings :", total_listings)
S_lim = 20
count_condition = (data["surface_n"]>S_lim).sum() 
print("Count of listings such that bigger than",S_lim,":", count_condition )
print("Percentage: %.2f" %   (count_condition / total_listings * 100))
data[data["surface_n"]>S_lim]
```

```python
avg_price_milan = 20
common_area = 0.2
threshold_value = 0.5

condition = data["surface_n"] * avg_price_milan > (1+ threshold_value) * data["price_n"]

# If condition is True, calculate surface_room. If False, keep existing or set to NaN/0
data["surface_room"] = np.where(condition, data["surface_n"]*(1-common_area) / data["rooms_n"], data["surface_n"])

data.head()
```

```python
total_listings = len(data["surface_room"])
print("Number of Listings :", total_listings)
S_lim = 25
count_condition = (data["surface_room"]>S_lim).sum() 
print("Count of listings such that bigger than",S_lim,":", count_condition )
print("Percentage: %.2f" %   (count_condition / total_listings * 100))
data[data["surface_room"]>S_lim]
```

```python
plt.hist(data["surface_room"],bins=40)
```

```python
total_listings = len(data["surface_room"])
print("Number of Listings :", total_listings)
S_lim = 30
count_condition = (data["surface_room"]>S_lim).sum() 
print("Count of listings such that bigger than",S_lim,":", count_condition )
print("Percentage: %.2f" %   (count_condition / total_listings * 100))
data[data["surface_room"]>S_lim]
```

```python
avg_room_size = 20
multiplier = 1.05

# Calculate IQR (Interquartile Range)
Q1 = data["surface_room"].quantile(0.25)
Q3 = data["surface_room"].quantile(0.75)
IQR = Q3 - Q1

# Define outliers as anything 1.5 times the IQR outside the boxes
statistical_outliers = (data["surface_room"] < (Q1 - multiplier * IQR)) | (data["surface_room"] > (Q3 + multiplier * IQR))

# Replace them
data.loc[statistical_outliers, "surface_room"] = avg_room_size
```

```python
total_listings = len(data["surface_room"])
print("Number of Listings :", total_listings)
S_lim = 25
count_condition = (data["surface_room"]>S_lim).sum() 
print("Count of listings such that bigger than",S_lim,":", count_condition )
print("Percentage: %.2f" %   (count_condition / total_listings * 100))
data[data["surface_room"]>S_lim]
```

```python
# Drop a list of columns
data_cleaned = data.drop(["rooms", "surface","floor"], axis=1)
data_cleaned.to_csv("./treated_surface.csv",index=False)
```
